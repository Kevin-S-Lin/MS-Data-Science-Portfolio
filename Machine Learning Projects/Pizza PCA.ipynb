```python
import pandas
import numpy
# Load Pizza.csv
data = pandas.read_csv('/content/Pizza.csv')
print(data.head())
```

## Solving for part (a)

```python
# Select the features
features = ['mois','prot','fat','ash','sodium','carb','cal']
# Calculate the mean vector
mean_vector = data[features].mean(axis=0)
print("Mean Feature Vector:")
print(mean_vector)
```

## Solving for part (b)

```python
# Get the scatter matrix also known as the covariance matrix
scatter_matrix = data[features].cov()
print("Scatter Matrix:")
print(scatter_matrix)
```

```python
# Calculate the the eigenvectors and eigenvalues of the scatter matrix.
# This corresponds to the direction of the principal components as well as their weight or how much variance they account for.
eigenvalues, eigenvectors = numpy.linalg.eig(scatter_matrix.values)
# Sort the eigenvectors by their eigenvalues to find the order of the principal components.
indices = numpy.argsort(eigenvalues)[::-1]
print("Eigenvalues")
eigenvalues = eigenvalues[indices]
print(eigenvalues)
print("\nEigenvectors")
eigenvectors = eigenvectors[:, indices]
print(eigenvectors)
```

```python
# Unit Length of principal components
ul_evector = eigenvectors / numpy.linalg.norm(eigenvectors, axis=0)
print(ul_evector)
```

```python
print("The first principal component:")
print(ul_evector[0])
print("\nThe second principal component:")
print(ul_evector[1])
```

## Solving for part (c)

```python
pc1 = ul_evector[0]
pc2 = ul_evector[1]
row1 = data[features].values[0]
row2 = data[features].values[1]
mean = mean_vector.to_numpy()
print("First two feature vectors:")
print("First feature vector:", row1)
print("Second feature vector:", row2)

print("\nApproximations:")
a_1 = (pc1.T).dot(row1 - mean)
a_2 = (pc2.T).dot(row1 - mean)
x_k = mean + a_1 * pc1 + a_2 * pc2
print("First approximation:", x_k)
a_1 = (pc1.T).dot(row2 - mean)
a_2 = (pc2.T).dot(row2 - mean)
x_k = mean + a_1 * pc1 + a_2 * pc2
print("Second approximation:", x_k)
```

## Solving for part (d)

```python
# First get standard deviation of each feature in the dataset
standard_deviation = data[features].std()
print("Standard Deviations:")
print(standard_deviation)
print("\n Standardized Dataset")
# Standardize the entire dataset
standardized_data = (data[features] - mean_vector) / standard_deviation
print(standardized_data)
```

```python
# Repeating part (b) of the problem
std_scatter = standardized_data.cov()
print("Standardized Scatter Matrix:")
print(scatter_matrix)

print("\n\nStandardized Principal Components 1 and 2")
std_eval, std_evec = numpy.linalg.eig(std_scatter.values)
indices = numpy.argsort(std_eval)[::-1]
std_eval = std_eval[indices]
std_evec = std_evec[:, indices]
print(std_evec[0])
print(std_evec[1])
```

```python
# Repeating part (c) of the problem
std_pc1 = std_evec[0]
std_pc2 = std_evec[1]
row1 = data[features].values[0]
row2 = data[features].values[1]
mean = mean_vector.to_numpy()
print("First two feature vectors:")
print("First feature vector:", row1)
print("Second feature vector:", row2)

print("\nApproximations:")
a_1 = (std_pc1.T).dot(row1 - mean)
a_2 = (std_pc2.T).dot(row1 - mean)
x_k = mean + a_1 * std_pc1 + a_2 * std_pc2
print("First approximation:", x_k)
a_1 = (std_pc1.T).dot(row2 - mean)
a_2 = (std_pc2.T).dot(row2 - mean)
x_k = mean + a_1 * std_pc1 + a_2 * std_pc2
print("Second approximation:", x_k)
```
