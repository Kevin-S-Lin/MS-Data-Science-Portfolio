## Problem 1

(3.1) For an MA(1), $x_t = w_t + \theta w_{t-1}$, show that $|\rho_x(1)| \leq 1/2$ for any number $\theta$. For which values of $\theta$ does $\rho_x(1)$ attain its maximum and minimum?

Useful Definition for MA(1):
$$
\rho(h) =
\begin{cases}
  \frac{\theta}{1+\theta^2} & h= 1 \\
  0 & h\gt 1
\end{cases}
$$

Let's try plugging in some random values to see if we can find a pattern. The goal is to see if the the absolute value of the autocorrelation is less than or equal to $1/2$ for any number $\theta$.

Let's try $\theta=0,0.5,1,2$
$$
\frac{0}{1+0^2}= 0 \\
\frac{0.5}{1+0.5^2} = \frac{0.5}{1.25} = 0.4\\
\frac{1}{1+1^2}= \frac{1}{2} = 0.5\\
\frac{2}{1+2^2}= \frac{2}{5} = 0.4
$$

We can see a clear pattern from numbers we tried. The denominator grows with the exponent of 2, while the numerator grows linearly.

We can also see that the autocorrelation reaches its minimum and maximum at $\theta=-1,1$ which gives $-1/2, 1/2$ respectively. As $\theta$ approaches infinity the autocorrelation approaches 0.

Addtionally, create a plot of $\rho_x(1)$ for values of $\theta$ ranging from -10 to 10. Comment on whether it confirms you answer to (3.1), and what you can say about reciprocal values for $\theta$ based on the plot.

```python
import plotly.express as px
import pandas as pd
import numpy as np

# Define the function for the autocorrelation of an MA(1) process at lag 1
def rho_x_1(theta):
  return theta / (1 + theta**2)

# Generate values for theta from -10 to 10
theta_values = np.linspace(-10, 10, 10000)

# Calculate the corresponding rho_x(1) values
rho_values = rho_x_1(theta_values)

# Create a DataFrame for Plotly Express
df = pd.DataFrame({'theta': theta_values, 'rho_x(1)': rho_values})

# Plot the function
fig = px.line(df, x='theta', y='rho_x(1)', title='Autocorrelation Function at Lag 1 for MA(1)')
fig.update_layout(xaxis_title='θ', yaxis_title='Autocorrelation')
fig.show()
```

The plot confirms my answer in (3.1). The absolute value peaks at $\theta = -1, 1$ like I thought at 1/2. Then it approaches 0 as the absolute value of $\theta$ goes to infinity.

(3.2, a-c) Let $\{ w_t;t=0,1,...\}$ be a white noise process with variance $\sigma_w^2$ and let $|\phi| \lt 1$ be a constant. Consider the process $x_0=w_0$, and
$$
x_t=\phi x_{t-1} + w_t,\ \ t=1,2,...
$$
We might use this method to simulate an AR(1) process from simulated white noise.

### Part (a)

Show that $x_t=\sum^t_{j=0}\phi^j w_{t-j}$ for any $t=0,1,...$

I believe if build it up from $x_0 = w_0$ we can begin to see a pattern.

We are given $x_0 = w_0$, so first lets plug that into the summation
$$
x_0 = \sum^0_{j=0}\phi^0w_{0-0} = (1)w_0=w_0
$$
<br>
We can see that it holds true for $x_0=w_0$.


---


Now let's attempt $x_1$
$$
x_1 = \phi x_{1-1} + w_1 = \phi x_{0} + w_1
$$
We can substitute $x_0$ since we already know it to be $w_0$
$$
x_1 = \phi w_0 + w_1
$$

Now let's try using the summation to the get $x_1 = \phi w_0 + w_1$
$$
x_1 = \sum^1_{j=0}\phi^jw_{1-j} \\
x_1 = \phi^0w_{1-0} + \phi^1w_{1-1} \\
x_1 = (1)w_{1} + \phi w_{0} = w_{1} + \phi w_{0} \\
x_1 = \phi w_0 + w_1
$$
It works.


---


Now let's try $x_2$
$$
x_2 = \phi x_{2-1} + w_2 = \phi x_1 + w_2 \\
x_2 = \phi (\phi w_0 + w_1) + w_2 \\
x_2 = \phi^2 w_0 + \phi w_1 + w_2
$$
And now, check to see if we get the same thing with the summation
$$
x_2 = \sum^2_{j=0}\phi^jw_{2-j} \\
x_2 = \phi^0w_{2-0} + \phi^1w_{2-1} + \phi^2 w_{2-2} \\
x_2 = (1)w_{2} + \phi w_{1} + \phi^2 w_{0} \\
x_2 = \phi^2 w_0 + \phi w_1 + w_2
$$
This also works and we can begin to see a pattern.

As the older terms get plugged in and expanded out, $\phi$ gradually increases in power for older timesteps of $w_t$. This is properly captured in the summation provided.

### Part (b)

Find the $E(x_t)$.

To find the expected value of $x_t$ we have enough information just plug in and solve.
$$
E(x_t) = E(\phi x_{t-1} + w_t) \\
E(x_t) = E(\phi x_{t-1}) + E(w_t) \\
E(x_t) = \phi E( x_{t-1}) + E(w_t)
$$
We know the mean of white noise is 0, so we can remove the last term to get:
$$
E(x_t) = \phi E( x_{t-1})
$$


---


And since we are given $x_0 =w_0$, we can try solving to for several timesteps to look for a pattern.
$$
E(x_0) = E(w_0) = 0 \\
E(x_1) = \phi E(x_{0}) = \phi (0) = 0 \\
E(x_2) = \phi E(x_1) = \phi (0) = 0 \\
$$


---

From the above we can see that $E(x_t)=0$ for all $t=0,1,2,...$

### Part (c)

Show that, for $t=0,1,...,$
$$
var(x_t) = \frac{\sigma_w^2}{1-\phi^2}(1-\phi^{2(t+1)})
$$

We can plug the original process in and try to solve $var(x_t)$
$$
var(x_t) = var(\phi x_{t-1} + w_t)\\
var(x_t) = \phi^2 var(x_{t-1}) + var(w_t) \\
var(x_t) = \phi^2 var(x_{t-1}) + \sigma_w^2
$$
The variance of $w_t$ is given in the problem so it is used to substitute it above.

---

Like the previous parts of this problem, let's try solving from the given $x_0 = w_0$
$$
var(x_0) = var(w_0) = \sigma_w^2
$$

---

Now, let's try $t=1$
$$
var(x_1) = \phi^2 var(x_{1-1}) + \sigma_w^2 \\
var(x_1) = \phi^2 var(x_{0}) + \sigma_w^2 \\
var(x_1) = \phi^2 \sigma_w^2 + \sigma_w^2 \\
var(x_1) = \sigma_w^2(\phi^2 + 1)
$$

---

Now, let's try $t=2$
$$
var(x_2) = \phi^2 var(x_{2-1}) + \sigma_w^2 \\
var(x_2) = \phi^2 var(x_{1}) + \sigma_w^2 \\
var(x_2) = \phi^2 \sigma_w^2(\phi^2 + 1) + \sigma_w^2 \\
var(x_2) = \sigma_w^2(\phi^2(\phi^2 + 1) + 1) \\
var(x_2) = \sigma_w^2(\phi^4 + \phi^2 + 1)
$$
From this we can see an emerging pattern.

With each timestep, we can see that $\sigma^2_w$ is a common term due to sitting outside the $var(x_{t-1})$ term. On the other hand, we can see that the terms with $\phi^2$ begin to pile up like a summation.

Representing $var(x_t)$ with a summation would look something like this
$$
var(x_t) = \sigma^2_w \sum^t_{j=0} \phi^{2j}
$$

Looking up the shortcut for a finite geometric sequence we get this:
$$
\sum^n_{k=0}r^k = \frac{1-r^{n+1}}{1-r}
$$

Applying this to our geometric sequence would yield:
$$
var(x_t) = \sigma^2_w \sum^t_{j=0} \phi^{2j} \\
var(x_t) = \sigma^2_w \frac{1-(\phi^2)^{t+1}}{1-\phi^2} \\
var(x_t) = \frac{\sigma^2_w}{1-\phi^2} (1-(\phi^2)^{t+1})
$$
And then we get the form of $var(x_t)$ that the question wants above by moving the term terms around a bit.

## Problem 3

(3.4) Identify the following models as ARMA(p,q) models (watch out for parameter redundancy), and determine whether they are causal and/or invertible:

### Part (a)

$x_t=.80x_{t-1} - .15x_{t-2} + w_t -.30w_{t-1}$

Largest lag of $x$ is 2, so $p=2$.

Largest lag of $w$ is 1, so $q=1$.

This gives ARMA(2,1).

Rewrite the equation with backshift operators
$$
x_t=.80x_{t-1} - .15x_{t-2} + w_t -.30w_{t-1} \\
x_t-.80x_{t-1} + .15x_{t-2}= w_t -.30w_{t-1} \\
(1-.80B + .15B^2)x_t= (1 -.30B)w_t
$$

This gives us the AR and MA polynomials $\phi(z)$ and $\theta(z)$
$$
\phi(z) = 1-.80z + .15z^2 \\
\theta(z) = 1 -.30z
$$

---

Check causality (roots must lie outside the unit circle)
$$
\phi(z) = 1-.80z + .15z^2 = 0
$$
Use quadratic formula in this case (I used a quadratic formula calculator).

We get $(|z|= \frac{10}{3}, 2)$ which are both outside of the unit circle, hence, the model is causal.

---

Check invertibility
$$
\theta(z) = 1 -.30z = 0 \\
1 = .30z \\
z = \frac{1}{.30} \approx 3.333
$$
The root has a magnitude greater than the unit circle, hence, the model is invertible.

---

Check parameter redundancy.

We see that both AR and MA polynomials share 3.333 so the model can be further simplified by removing $(1-.3B)$.

### Part (b)

$x_t=x_{t-1}-.50x_{t-2}+w_t-w_{t-1}$

Largest lag of $x$ is 2, so $p=2$.

Largest lag of $w$ is 1, so $q=1$.

This give ARMA(2,1).

Rewrite the equation with backshift operators
$$
x_t=x_{t-1}-.50x_{t-2}+w_t-w_{t-1} \\
x_t-x_{t-1}+.50x_{t-2}=w_t-w_{t-1} \\
(1-B+.50B^2)x_t=(1-B)w_t
$$

This gives us the AR and MA polynomials $\phi(z)$ and $\theta(z)$
$$
\phi(z) = 1-z + .50z^2 \\
\theta(z) = 1 -z
$$

---

Check causality (roots must lie outside the unit circle)
$$
\phi(z) = 1-z + .50z^2 = 0
$$
Use quadratic formula in this case (I used a quadratic formula calculator).

We get $(z= 1 \pm i)$ which when converted into magnitude is $(|z| = \sqrt{2})$. Hence, with both roots greater than 1 and outside of the unit circle, the model is causal.

---

Check invertibility (roots must lie outside the unit circle)
$$
\theta(z) = 1 -z = 0 \\
z = 1
$$
The root does not have a magnitude greater than the unit circle, hence, the model is not invertible.

---

Check parameter redundancy.

No roots are shared so there is no parameter redundancy.

## Problem 4

(3.6) For the AR(2) model given by $x_t=-.9x_{t-2} +w_t$, find the roots of the autoregressive polynomial, and then plot the ACF, $\rho(h)$.

First, rewrite the equation with backshift operators
$$
x_t=-.9x_{t-2} +w_t \\
x_t+.9x_{t-2}= w_t \\
(1+.9B^2)x_t= w_t
$$

Next, get AR from $\phi(z)$
$$
\phi(z) = 1+.9z^2
$$

Find the roots
$$
\phi(z) = 1+.9z^2 = 0 \\
z^2 = -\frac{1}{.9} \\
z = \pm i \sqrt{\frac{1}{.9}} = \pm i \frac{\sqrt{10}}{3}
$$

Now, get the magnitudes
$$
|z| = \sqrt{0^2 + (\frac{\sqrt{10}}{3})^2} = \frac{\sqrt{10}}{3}
$$
The roots are greater than 1 and hence lie outside the unit circle. This means the AR(2) model is causal.

```python
# Now to plot the ACF, rho(h)
import plotly.express as px
import numpy as np
from statsmodels.tsa.arima_process import ArmaProcess

# Define the ARMA process (taken from class example code)
ar2_process = ArmaProcess(ar=[1, 0, -0.9],
                          ma=[1],
                          )

# Print the coefficients
print("Coefficients:",ar2_process.arcoefs)

# Print the roots
print("Roots:",ar2_process.arroots)

# Now get the ACF
rho_h = ar2_process.acf(lags=100)

# Plot the ACF using plotly
fig = px.line(x=np.arange(len(rho_h)), y=rho_h, title='Autocorrelation Function')
fig.update_layout(xaxis_title='Lag', yaxis_title='Autocorrelation')
fig.show()
```
