# Problem 2

Find the autocovariance and autocorrelation for $s=t,|s-t|=1,|s-t|=2,\ and\ |s-t|=3$ of the stationary moving average process $v_t=\frac{1}{4}w_{t-1}+\frac{1}{2}w_t + \frac{1}{4}w_{t+1}$ for iid $w_t$ with $w_t \sim wn(0,\sigma^2_w)$.

Useful Definitions:
$$
\gamma_x(s, t) = E[(x_s-\mu_s)(x_t-\mu_t)] =E[x_s x_t]-\mu_s\mu_t \\\\
E[w_tw_s] = \sigma_w^2\ and\ 0\ otherwise \\\\
\rho(s, t) = \frac{\gamma(s,t)}{\sqrt{\gamma(s,s)\gamma(t,t)}}
$$

Mean is 0 so the moving average does not change the mean. This means we can set the expected value of $v_t$ to 0.
$$
E[v_t] = E[\frac{1}{4}w_{t-1}+\frac{1}{2}w_t + \frac{1}{4}w_{t+1}] \\\\
E[v_t] = \frac{1}{4}E[w_{t-1}]+\frac{1}{2}E[w_t] + \frac{1}{4}E[w_{t+1}] \\\\
E[v_t] = \frac{1}{4}E[w_{t-1}]+\frac{1}{2}E[w_t] + \frac{1}{4}E[w_{t+1}] = 0\\\\
$$

And mean is 0, so when plugged into the autocovariance formula, we get:
$$
\gamma_x(s, t) =E[x_s x_t]-\mu_s\mu_t \\\\
\gamma_x(s, t) =E[x_s x_t]-0 \\\\
\gamma_x(s, t) =E[x_s x_t]\\\\
$$

Solving for autocovariance:

This gives us $E[v^2_t]$ for $s=t$ (expected value of cross-terms is 0)
$$
E[v^2_t] = E[(\frac{1}{4}w_{t-1}+\frac{1}{2}w_t + \frac{1}{4}w_{t+1})^2] \\\\
E[v^2_t] = E[\frac{1}{16}w^2_{t-1}+\frac{1}{4}w^2_t + \frac{1}{16}w^2_{t+1}] \\\\
E[v^2_t] = \frac{1}{16}E[w^2_{t-1}]+\frac{1}{4}E[w^2_t] + \frac{1}{16}E[w^2_{t+1}]  \\\\
E[v^2_t] = \frac{1}{16}\sigma^2_w+\frac{1}{4}\sigma^2_w + \frac{1}{16}\sigma^2_w \\\\
E[v^2_t] = \frac{3}{8}\sigma^2_w \\\\
$$

And $E[v_tv_{t-1}]$ for $|s-t|=1$
$$
E[v_tv_{t-1}] = E[(\frac{1}{4}w_{t-1}+\frac{1}{2}w_t + \frac{1}{4}w_{t+1})(\frac{1}{4}w_{t-2}+\frac{1}{2}w_{t-1} + \frac{1}{4}w_{t})] \\\\
E[v_tv_{t-1}] = E[(\frac{1}{2}w_{t})(\frac{1}{4}w_t)+(\frac{1}{4}w_{t-1})(\frac{1}{2}w_{t-1})] \\\\
E[v_tv_{t-1}] = \frac{1}{8}E[w_{t}] + \frac{1}{8}E[w_{t-1}]\\\\
E[v_tv_{t-1}] = \frac{1}{8}\sigma^2_w + \frac{1}{8}\sigma^2_w\\\\
E[v_tv_{t-1}] = \frac{1}{4}\sigma^2_w\\\
$$

And $E[v_tv_{t-2}]$ for $|s-t|=2$
$$
E[v_tv_{t-2}] = E[(\frac{1}{4}w_{t-1}+\frac{1}{2}w_t + \frac{1}{4}w_{t+1})(\frac{1}{4}w_{t-3}+\frac{1}{2}w_{t-2} + \frac{1}{4}w_{t-1})] \\\\
E[v_tv_{t-2}] = E[(\frac{1}{4}w_{t-1})(\frac{1}{4}w_{t-1})] \\\\
E[v_tv_{t-2}] = \frac{1}{16}E[w_{t-1}^2] \\\\
E[v_tv_{t-2}] = \frac{1}{16}\sigma^2_w \\\\
$$

And $E[v_tv_{t-3}]$ for $|s-t|=3$
$$
E[v_tv_{t-3}] = E[(\frac{1}{4}w_{t-1}+\frac{1}{2}w_t + \frac{1}{4}w_{t+1})(\frac{1}{4}w_{t-4}+\frac{1}{2}w_{t-3} + \frac{1}{4}w_{t-2})] \\\\
E[v_tv_{t-3}] = 0 \\\\
$$

Solving for autocorrelation:

For $(s=t)$:
$$
\rho(0)=\frac{\gamma(0)}{\gamma(0)} = 1
$$

For $(|s-t|=1)$:
$$
\rho(1)=\frac{\gamma(1)}{\gamma(0)} = \frac{\frac{1}{4}\sigma^2_w}{\frac{3}{8}\sigma^2_w} = \frac{2}{3}
$$

For $(|s-t|=2)$:
$$
\rho(2)=\frac{\gamma(2)}{\gamma(0)} = \frac{\frac{1}{16}\sigma^2_w}{\frac{3}{8}\sigma^2_w} = \frac{1}{6}
$$

For $(|s-t|=3)$:
$$
\rho(3)=\frac{\gamma(3)}{\gamma(0)} = \frac{0}{\frac{3}{8}\sigma^2_w} = 0
$$

## Problem 1.6

Consider the time series
$$
x_t = \beta_1 + \beta_2 t + w_t
$$
where $\beta_1$ and $\beta_2$ are known constants with $w_t$ is a white noise process with variance $\sigma^2_w$.

### Part (a)

Determine whether $x_t$ is stationary.

A time series needs a constant mean and variance over time in order to be stationary. To check for this I'll try to calculate the mean of $x_t$.
$$
E[x_t] = E[\beta_1 + \beta_2t + w_t] \\\\
E[x_t] = \beta_1 + \beta_2t + E[w_t] \\\\
E[x_t] = \beta_1 + \beta_2t \\\\
$$

The $E[w_t]$ is removed because we know the mean of white noise is 0. This result tells us that that the mean of $x_t$ is $\beta_1 + \beta_2t$ and therefore depends on time because of the $t$ term. Hence, the mean is not constant and the time series is not stationary.

### Part (b)

Show that the process $y_t=x_t - x_{t-1}$ is stationary.

Check for stationarity conditions

First, let's simplify:
$$
y_t=x_t - x_{t-1}\\\\
y_t=(\beta_1 + \beta_2 t + w_t) - (\beta_1 + \beta_2(t-1) + w_{t-1})\\\\
y_t= \beta_2 + w_t - w_{t-1}\\\\
$$

Now, check for constant mean:
$$
E[y_t] = E[\beta_2 + w_t - w_{t-1}]\\\\
E[y_t] = \beta_2 + E[w_t] - E[w_{t-1}]\\\\
E[y_t] = \beta_2\\\\
$$
Mean is constant at $\beta_2$ (mean of white noise is 0, hence removed)

Now, check for constant variance:
$$
Var(y_t) = Var(\beta_2 + w_t - w_{t-1})\\\\
Var(y_t) = Var(w_t) + Var(-w_{t-1})\\\\
Var(y_t) = \sigma_w^2 + \sigma_w^2 = 2\sigma_w^2\\\\
$$
Variance is constant at $2\sigma_w^2$ ($\beta_2$ is removed due to being a constant)

Now, check for constant autovariance ($h$ is lag):
$$
E[(y_t - E[y_t])(y_{t-h} - E[y_{t-h}])]=E[(w_t-w_{t-1})(w_{t-h}-w_{t-h-1})]
$$
At $h=1$:
$$
E[(w_t-w_{t-1})(w_{t-1}-w_{t-2})] = E[-w^2_{t-1}] = -\sigma^2_w
$$
Constant mean with with no more common white noise terms beyond $h=1$, hence covariance is zero for $h \geq 2$. Autocovariance only depends on lag and not time.

Mean, variance, and covariance are all constant and independent of time, therefore $y_t$ is stationary.

### Part (a)

Generate four series that are random walk with drift, (1.4), of length $n = 100$ with $δ = .01$ and $σ_w = 1$. Call the data $x_t$ for $t = 1, . . . , 100$. Fit the regression $x_t = βt + w_t$ using least squares. Plot the data, the true mean function (i.e.,
$µ_t = .01 t$) and the fitted line, $x̂_t = β̂ t$, on the same graph. Hint: The following R code may be useful.

```python
# Common parameters
n = 100
delta = 0.01
sigma = 1
```

```python
import numpy as np
import plotly.graph_objects as go

# Random Walk with Drift Function
def random_walk_with_drift(n, delta, sigma):
    cumsum = np.random.normal(loc=delta, scale=sigma, size=n).cumsum()
    cumsum = np.insert(cumsum, 0, 0)
    return cumsum
```

```python
# Generate four series using random walk function
x1 = random_walk_with_drift(n, delta, sigma)
x2 = random_walk_with_drift(n, delta, sigma)
x3 = random_walk_with_drift(n, delta, sigma)
x4 = random_walk_with_drift(n, delta, sigma)

# Plot the four random walks with a true mean line and regression line
t = np.arange(n + 1)
true_mean = delta * t

fig = go.Figure()

fig.add_trace(go.Scatter(x=t, y=x1, mode='lines', name='Random Walk 1'))
fig.add_trace(go.Scatter(x=t, y=x2, mode='lines', name='Random Walk 2'))
fig.add_trace(go.Scatter(x=t, y=x3, mode='lines', name='Random Walk 3'))
fig.add_trace(go.Scatter(x=t, y=x4, mode='lines', name='Random Walk 4'))
fig.add_trace(go.Scatter(x=t, y=true_mean, mode='lines', name='True Mean', line=dict(color='black', dash='dash')))

# Fit regression line and add to plot
# For each random walk, fit a regression line and add it to the plot
for i, x in enumerate([x1, x2, x3, x4]):
    # Fit linear regression
    # Drop the first element which is always 0 to fit the model x_t = beta*t + w_t
    x_fit = x[1:]
    t_fit = t[1:]
    beta_hat = np.sum(t_fit * x_fit) / np.sum(t_fit**2)
    fitted_line = beta_hat * t

    fig.add_trace(go.Scatter(x=t, y=fitted_line, mode='lines', name=f'Fitted Line {i+1}', line=dict(dash='dot')))


fig.update_layout(title='Random Walk with Drift vs. True Mean and Fitted Lines',
                  xaxis_title='Time (t)',
                  yaxis_title='x_t')
fig.show()
```

## Problem 2.6

Consider a process consisting of a linear trend with an additive noise term consisting of independent random variables $w_t$ with zero means and variances $σ_w^2$ , that
is,
$$
x_t = \beta_0 + \beta_1 t + w_t
$$
where $\beta_0\beta_1$ are fixed constants.

### Part (a)

Prove $x_t$ is nonstationary.

Check for stationarity

Check for constant mean
$$
E[x_t] = E[\beta_0 + \beta_1 t + w_t] \\\\
E[x_t] = \beta_0 + \beta_1 t + E[w_t] \\\\
E[x_t] = \beta_0 + \beta_1 t \\\\
$$
Since the mean depends on time, the process is not stationary.

### Part (b)

Prove that the first difference series $∇x_t = x_t - x_{t−1}$ is stationary by finding its mean and autocovariance function.

First, simplify
$$
∇x_t = x_t - x_{t−1} \\\\
∇x_t = (\beta_0 + \beta_1 t + w_t) - (\beta_0 + \beta_1 (t-1) + w_{t-1}) \\\\
∇x_t = \beta_1 + w_t  - w_{t-1}
$$

Now, check for constant mean
$$
E[∇x_t] = E[\beta_1 + w_t  - w_{t-1}] \\\\
E[∇x_t] = \beta_1 + E[w_t]  - E[w_{t-1}] \\\\
E[∇x_t] = \beta_1
$$
The mean is constant at $\beta_1$

Now, check for time dependence in autocovariance
$$
\gamma(h) = E[(∇x_t - \beta_1)(∇x_{t-h} - \beta_1)]\\\\
\gamma(h) = E[(w_t - w_{t-1})(w_{t-h} - w_{t-h-1})]
$$

Solve of lag $h=0$
$$
\gamma(0) = E[(w_t - w_{t-1})(w_t - w_{t-1})]\\\\
\gamma(0) = E[w_t^2 - 2w_tw_{t-1} + w^2_{t-1}] \\\\
\gamma(0) = \sigma_w^2 -0 + \sigma_w^2 = 2\sigma_w^2
$$

Solve for lag $h=1$
$$
\gamma(1) = E[(w_t - w_{t-1})(w_{t-1} - w_{t-2})]\\\\
\gamma(1) = E[-w_{t-1}^2] = -\sigma_w^2\\\\
$$

For $|h|\geq 2$ there are no more shared noise terms so the autocovariance is 0.

The first difference series passes the checks for constant mean and an autocovariance that doesn't depend on time. Hence, $∇x_t$ is stationary.

### Part (c)

Repeat part (b) if $w_t$ is replaced by a general stationary process, say $y_t$ , with mean function $µ_y$ and autocovariance function $γ_y (h)$.

\gamma(h) = E[w_tw_{t-h}] + E[y_ty_{t-h}] - E[y_{t-1}y_{t-h}] - E[y_ty_{t-h-1}] + E[y_{t-1}y_{t-h-1}]
$$
This can be rewritten as
$$
\gamma(h) = Cov(w_t,w_{t-h}) + Cov(y_t,y_{t-h}) - Cov(y_{t-1},y_{t-h}) - Cov(y_t,y_{t-h-1}) + Cov(y_{t-1},y_{t-h-1})
$$
From this we can see that the autocovariance only depends on the lag and not on time.

In conclusion, mean is constant and autocovariance is independent of time meaning that $x_t - x_{t-1}$ is stationary.
