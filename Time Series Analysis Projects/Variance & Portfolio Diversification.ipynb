# Problem 1

We have seen that the variance of the sum of two random variables can be greater than the sum of each variable's variance if the variables have a positive covariance. Nevertheless, financial advisers, who are generally more concerned with the standard deviation than the variance, still advise clients to diversify as long as the assets are not perfectly correlated. Assuming the expected returns are the same, demonstrate why, when considering the standard deviation, it is better to split an investment between two assets rather than putting everything into a single asset. Hint: start with the maximum possible standard deviation of the sum of two random variables and reason from there.

Variance of the sum of 2 variables:
$$
V(X+Y)=V(X)+V(Y)+2Cov(X,Y)
$$
Variance-Covariance Inequalities
$$
|Cov(X,Y)|\leq \frac{1}{2}(V(X) + V(Y)) \\\\
|Cov(X,Y)|\leq \sqrt{V(X)  V(Y)}
$$
<br>

Maximum possible standard deviation of the sum of two random variables:
$$
\sigma^{max}_{X+Y} = \sqrt{V(X+Y)}\\\\
\sigma^{max}_{X+Y} = \sqrt{V(X)+V(Y)+2Cov(X,Y)}\\\\
\sigma^{max}_{X+Y} = \sqrt{V(X)+V(Y)+2\sqrt{V(X)  V(Y)}} \\\\
\sigma^{max}_{X+Y} = \sqrt{ (\sqrt{V(X)} + \sqrt{V(Y)} )^2 } \\\\
\sigma^{max}_{X+Y} = \sqrt{V(X)} + \sqrt{V(Y)} \\\\
\sigma^{max}_{X+Y} = \sigma_X + \sigma_Y
$$

In the case that covariance is maximized, this would mean that the two assets are perfectly correlated. Conversely, if they aren't perfectly correlated, we get:
$$
\sigma_{X+Y} < \sigma_X + \sigma_Y
$$

Hence, as long as the two assets aren't perfectly correlated the expected return stays the same while standard deviation drops since it would have to be less than the sum of the standard deviation of the two separate assets. All in all, diversifying assets lowers the portfolio's cumulative standard deviation (if not perfectly correlated) which means less volatility.
