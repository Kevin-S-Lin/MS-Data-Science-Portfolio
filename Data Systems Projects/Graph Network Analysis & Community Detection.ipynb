# Graph Design

## Migrating ComfyMug Coffee Roasters

To take the ComfyMug relational database and convert all of it to a property graph model is possible. The main point is to turn each row into it's own node and each column into properties of that node.

The relations on the other hand stay fairly similar:
- `(customer--have->orders)`
- `(orders--draws_from->inventory)`
- `(farmers--supply->inventory)`

The original floating table of `Employees` would also just translate into floating Nodes with properties but no relations.

The joining table `OrderItems` is no longer necessary because of being able to have properties on edges. Previously in the relational database I would need a table `OrderItems` to sit between `Orders` and `Inventory` because each row in `Orders` could have multiple `Inventory` items and each item would be represented as a row in `OrderItems` with 2 Foreign Keys. With PGM, an `order` node can just point to multiple `inventory` nodes with a property `quantity` and `roast` on each edge so no messy data structures need to be kept on each `order` node.

As for properties of edges:
- `have`
  - No Properties
- `draws_from`
  - `quantity`
  - `roast`
- `supply`
  - No Properties

```python
# I made a mistake and tried using the internal node ID as identification for order, now changed to uuid
import uuid

def generate_uuid():
    return str(uuid.uuid4())

driver = GraphDatabase.driver(uri, auth=(username, password))

driver.verify_connectivity()

# Sample order data
orders_data = [
    ("alice.johnson@example.com", '2025-02-10', [("Colombian Supremo", 2, True), ("Ethiopian Durato Bombe", 1, False)]),  # Customer 1, 2 lbs of Colombian Roasted, 1 lb of Ethiopian Raw
    ("bob.williams@example.com", '2025-02-10', [("Brazil Alta Mogiana", 3, True)]),        # Customer 2, 3 lbs of Brazil Roasted
    ("charlie.brown@example.com", '2025-02-10', [("Kenyan AA", 1, True), ("Guatemalan Huehuetenango", 2, True)]), # Customer 3, 1 lb of Kenyan, 2 lbs of Guatemalan
    ("david.lee@example.com", '2025-02-10', [("Colombian Supremo", 1, False)]),        # Customer 4, 1 lb of Colombian Raw
    ("eve.davis@example.com", '2025-02-10', [("Ethiopian Durato Bombe", 2, False), ("Brazil Alta Mogiana", 1, False)])   # Customer 5, 2 lbs of Ethiopian, 1 lb of Brazil
]

# Queries for getting the price of beans
roastedquery = """
            MATCH (i:Inventory)
            WHERE i.coffeetype = $coffee_key
            RETURN i.ppproasted AS price
        """

rawquery = """
            MATCH (i:Inventory)
            WHERE i.coffeetype = $coffee_key
            RETURN i.pppraw AS price
        """

# Loop through orders, create edges from orders nodes to inventory nodes, and customer nodes to order nodes, and find out the total price of an order
for email, order_date, order_items in orders_data:

    # Create Order node and make an edge pointing from the customer node to the order node
    order_id = generate_uuid()
    with driver.session() as session:
        session.execute_write(lambda tx: tx.run("MATCH (c:Customer) WHERE c.email = $email CREATE (c)-[:have]->(o:Order {id: $order_id, orderdate: $order_date})", email=email, order_id=order_id, order_date=order_date))

    # Get the total price of the order while creating the edge
    total_price = 0
    for coffee_key, quantity, roasted in order_items:

      # Query the cost of a coffee
      with driver.session() as session:

          # Get price
          if roasted:
            price = session.execute_read(lambda tx: tx.run(roastedquery, coffee_key=coffee_key).single()[0])
          else:
            price = session.execute_read(lambda tx: tx.run(rawquery, coffee_key=coffee_key).single()[0])

          # Create the edge Order -> Inventory, and add properties to the edge
          with driver.session() as session:
            session.execute_write(lambda tx: tx.run("MATCH (o:Order {id: $order_id}), (i:Inventory {coffeetype: $coffee_key}) CREATE (o)-[k:draws_from]->(i) SET k.quantity = $quantity, k.roasted = $roasted", order_id=order_id, coffee_key=coffee_key, quantity=quantity, roasted=roasted))

          total_price += price * quantity
          print(f"Total price for order {order_id}: ${total_price}")

    # Set the total price for the current order
    with driver.session() as session:
        session.execute_write(lambda tx: tx.run("MATCH (o:Order) WHERE id(o) = $order_id SET o.totalprice = $total_price", order_id=order_id, total_price=total_price))
```

# Graph Query

What question do I want to answer that leverages the effect of a Graph Database?
I think finding the most popular coffee and the least popular coffee effectively leverages the ability of a Graph Database to directly look at the number of edges pointing at a given node (ie. hot spots and cold spots).


Why does it matter to my business?
This matters because by finding out this information, I can get a better idea of future trends and how much to stock of each coffee.

How does my query make use of the fact it is a graph?
This leverages a Graph Database's power to see hot spots and cold spots via edges.

```python
driver = GraphDatabase.driver(uri, auth=(username, password))

driver.verify_connectivity()

# Query to find the most popular coffee
hotquery = """
          MATCH (o:Order)-[r:draws_from]->(i:Inventory)
          WITH i, count(r) AS timesOrdered
          ORDER BY timesOrdered DESC
          RETURN i.coffeetype, timesOrdered
          LIMIT 1;
        """

# Query to find the least popular coffee
coldquery = """
          MATCH (o:Order)-[r:draws_from]->(i:Inventory)
          WITH i, count(r) AS timesOrdered
          ORDER BY timesOrdered ASC
          RETURN i.coffeetype, timesOrdered
          LIMIT 1;
        """

def run_query(tx, query):
        result = tx.run(query)
        return result.data()

with driver.session() as session:
    resulthot = session.execute_read(run_query, hotquery)
    resultcold = session.execute_read(run_query, coldquery)

    print(f'The most popular coffee is {resulthot} with {resulthot} orders.')
    print(f'The least popular coffee is {resultcold} with {resultcold} orders.')

driver.close()
```

## Centrality

Degree Centrality as defined in the Lectures is the number of edges/connections a node has. A way to do this would be to query for Artist nodes and sort by the amount of connections that point towards it with the edge LISTEN_TO in order to find the most popular artist.
