(1) A die D has four sides with the following probabilities :
p(1) = p(2) = 1/8, p(3) = 1/4, p(4)= 1/2.
What is the entropy of this distribution?

(A) Calculate the entropy analytically usng the formula

(B) Calculate the entropy using Python code

(A) Entropy calculated analytically (This just means manually right?).

Formula: Entropy = - ∑ p(x) * logbase2(p(x))

Written out this would give us:

Entropy = - ( (1/8)logbase2(1/8) + (1/8)logbase2(1/8) + (1/4)logbase2(1/4) + (1/2)logbase2(1/2) )

Entropy = - ( (1/8)logbase2(2^-3) + (1/8)logbase2(2^-3) + (1/4)logbase2(2^-2) + (1/2)logbase2(2^-1) )

Entropy = - ( (1/8)(-3) + (1/8)(-3) + (1/4)(-2) + (1/2)(-1) )

Entropy = - ( -(6/8) - (2/4) - (1/2) )

Entropy = - ( -(6/8) - (4/8) - (4/8) ) = - ( -(14/8) ) = 7/4

Entropy = 7/4 or 1.75

```
from math import log2

# (B) Entropy calculated using python code.
H = -sum([p * log2(p) for p in [1/8, 1/8, 1/4, 1/2]])
H
```

(2) Consider the die D of quesion (1):

(A) Which probability distribution of the four sides will maximize the entropy, i.e., find the values of P(1), P(2), P(3), and P(4)?

(B) What is the value of the maximum entropy?

```
# p(1)=? , p(2)=?, p(3)=?, p(4)=?
# (A) Entropy is maximized when the probability distribution is Uniform hence, 1/4 for a four sided die.

# (B) The value of maximum entropy calculated using python.
p = [1/4, 1/4, 1/4, 1/4]
H = -sum([p_i * log2(p_i) for p_i in p])
H
```

(3) A coin C1 has the probability distribution p(T) = p(H) = 1/2
Another coin C2 has the probability distribution p(T) = 3/4, p(H) = 1/4
What is the relative entropy (Kullback–Leibler divergence ) from C2 to C1?
What is the relative entropy (Kullback–Leibler divergence ) from C1 to C2?

```
from scipy.stats import entropy
print("Relative Entropy for C2 to C1")
print(entropy([1/2, 1/2], qk=[3/4, 1/4], base=2))

print("\nRelative Entropy for C1 to C2")
print(entropy([3/4, 1/4], qk=[1/2, 1/2], base=2))
```

(4) Suppose we have a source with 5 symbols. The corresponding probabilities are (0.3,0.3,0.2,0.1,0.1).

(A) What is the best possible average bits per symbol if we want to compress the source?

(B) Give a Huffman coding for the source symbols and calculate the average bits/symbol?

```
a = [0.3, 0.3, 0.2, 0.1, 0.1]

print("Best possible average bits per symbol:")
print(-sum([p * log2(p) for p in a]))

print("\nHuffman Coding for the source symbols:")
print("Symbol a1 (0.3): 0, 1 bit")
print("Symbol a2 (0.3): 10, 2 bits")
print("Symbol a3 (0.2): 110, 3 bits")
print("Symbol a4 (0.1): 1110, 4 bits")
print("Symbol a5 (0.1): 1111, 4 bits")

print("\nAverage bits/symbol:")
print(0.3 * 1 + 0.3 * 2 + 0.2 * 3 + 0.1 * 4 + 0.1 * 4)
```
